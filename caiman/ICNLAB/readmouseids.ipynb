```python
#"C:\Users\ICNLab\caiman_data\CensusData\INTRSECT Project.xlsx" Read mouse ids here and print to list
import pandas as pd

# Path to your Excel file
excel_path = r"C:\Users\ICNLab\caiman_data\CensusData\INTRSECT Project.xlsx"

# Read Excel file
df = pd.read_excel(excel_path)

# Get first column, drop NaNs, convert to list
values = df.iloc[:, 0].dropna().tolist()

# Print to verify
print(values)

```

```python
import os
import shutil

# ===== USER SETTINGS =====
source_root = r"Y:\DaVinci" #search here
dest_root = r"Z:\INTRSECT_INVIVO" #copy here

folder_names = values
# =========================

# Make destination root if it doesn't exist
os.makedirs(dest_root, exist_ok=True)

copied = []
missing = []

for name in folder_names:
    src_dir = os.path.join(source_root, name)

    if os.path.isdir(src_dir):
        dst_dir = os.path.join(dest_root, name)

        print(f"Copying: {src_dir} → {dst_dir}")
        shutil.copytree(src_dir, dst_dir, dirs_exist_ok=True)

        copied.append(name)
    else:
        print(f"NOT FOUND: {src_dir}")
        missing.append(name)

print("\n===== SUMMARY =====")
print("Copied folders:", copied)
print("Missing folders:", missing)
```
